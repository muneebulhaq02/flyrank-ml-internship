## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The business objective is to prioritize pages for review. Since the starter dataset does not contain a direct review-priority label, a proxy target called is_declining_label is created from trend_direction, where pages labelled "down" are treated as positive examples. This proxy enables supervised learning while keeping the overall objective focused on ranking content for manual review.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Create proxy target used in the starter pipeline
df["is_declining_label"] = (df["trend_direction"] == "down").astype(int)

print("Target:", "is_declining_label")
print()

print(df["is_declining_label"].value_counts())

print()

print("Positive Rate:",
      round(df["is_declining_label"].mean(),3))
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

These factors interact in complex ways, making it difficult to capture useful patterns with simple if-statements. Machine learning can learn relationships among multiple features and produce a better ranking of pages for review. The model supports human decision-making rather than replacing it.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
important_features = [
    "impressions_90d",
    "ctr",
    "avg_position",
    "content_age_days",
    "sessions_90d",
    "engagement_rate"
]

print("Example features used for ranking:")

for feature in important_features:
    print("-", feature)

print()
print("Multiple interacting signals motivate using ML instead of one fixed rule.")
```
